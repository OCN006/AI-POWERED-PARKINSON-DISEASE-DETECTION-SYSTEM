```python
import wfdb
import matplotlib.pyplot as plt
import os

# Point to the downloaded directory structure
data_dir = r"D:\ParkinsonDisease\Sleepdata\physionet.org\files\bidsleep-dataset\1.0.0"

# Replace 'record_name' with the actual name of a data file in the folder (e.g., 'tr03-0005')
# Do not include the .dat or .hea extension
record_path = os.path.join(data_dir, "record_name")

# Read the record
record = wfdb.rdrecord(record_path)

# Print basic information about the signals
print(f"Signals: {record.sig_name}")
print(f"Sampling frequency: {record.fs} Hz")

# Plot the first few seconds of the waveforms
wfdb.plot_wfdb(record, title="Raw Sleep Signals")
```
